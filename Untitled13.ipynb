```python
import pandas as pd
import numpy as np

# Tạo dữ liệu giả
np.random.seed(0)

columns = [
    'T1','L1','H1','S1','V1','X1','D1','N1',
    'T2','L2','H2','S2','V2','X2','D2','N2',
    'T6','L6','H6','S6','V6','X6','D6','N6',
    'GT','DT','KV','KT','DH1','DH2','DH3'
]

data = np.random.uniform(0, 10, size=(100, len(columns)))
df = pd.DataFrame(data, columns=columns)

# Thêm dữ liệu định tính
df['GT'] = np.random.choice(['Nam', 'Nu'], size=100)
df['DT'] = np.random.choice([1,2,3,np.nan], size=100)
df['KV'] = np.random.choice([1,2,3], size=100)
df['KT'] = np.random.choice(['A','A1','B','C','D'], size=100)

# Lưu file
df.to_csv("dulieuxettuyendaihoc.csv", index=False)

print("Đã tạo file dulieuxettuyendaihoc.csv")
```

```python
import pandas as pd
import numpy as np

# 1. Đọc dữ liệu
df = pd.read_csv("dulieuxettuyendaihoc.csv")

# 2. Hiển thị 10 dòng đầu và cuối
print("10 dòng đầu:")
print(df.head(10))

print("\n10 dòng cuối:")
print(df.tail(10))
```

```python
# Thống kê dữ liệu thiếu
print("Thiếu DT:", df['DT'].isnull().sum())

# Tần số + tần suất
freq_DT = df['DT'].value_counts(dropna=False)
ratio_DT = df['DT'].value_counts(normalize=True, dropna=False)

print("Tần số:\n", freq_DT)
print("Tần suất:\n", ratio_DT)

# Giá trị riêng biệt
print("Unique DT:", df['DT'].unique())

# Điền giá trị thiếu = 0
df['DT'] = df['DT'].fillna(0)
```

```python
print("Thiếu T1:", df['T1'].isnull().sum())

# Tần số
print(df['T1'].value_counts(dropna=False))
print(df['T1'].value_counts(normalize=True, dropna=False))

# Điền bằng mean
mean_T1 = df['T1'].mean()
df['T1'] = df['T1'].fillna(mean_T1)
```

```python
score_cols = [col for col in df.columns if col.startswith(('T','L','H','S','V','X','D','N','DH'))]

for col in score_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mean())
```

```python
df['TBM1'] = (df['T1'] + df['L1'] + df['H1'] + df['S1'] + df['V1'] + df['X1'] + df['D1'] + df['N1']) / 8
df['TBM2'] = (df['T2'] + df['L2'] + df['H2'] + df['S2'] + df['V2'] + df['X2'] + df['D2'] + df['N2']) / 8
df['TBM3'] = (df['T6'] + df['L6'] + df['H6'] + df['S6'] + df['V6'] + df['X6'] + df['D6'] + df['N6']) / 8
```

```python
def classify(x):
    if x < 5:
        return 'Y'
    elif x < 6.5:
        return 'TB'
    elif x < 8:
        return 'K'
    elif x < 9:
        return 'G'
    else:
        return 'XS'

df['XL1'] = df['TBM1'].apply(classify)
df['XL2'] = df['TBM2'].apply(classify)
df['XL3'] = df['TBM3'].apply(classify)
```

```python
def classify(x):
    if x < 5:
        return 'Y'
    elif x < 6.5:
        return 'TB'
    elif x < 8:
        return 'K'
    elif x < 9:
        return 'G'
    else:
        return 'XS'

df['XL1'] = df['TBM1'].apply(classify)
df['XL2'] = df['TBM2'].apply(classify)
df['XL3'] = df['TBM3'].apply(classify)
```

```python
def kqxt(row):
    if row['KT'] in ['A', 'A1']:
        score = (row['DH1'] + row['DH2'] + row['DH3']) / 3
    elif row['KT'] == 'B':
        score = (row['DH1'] + row['DH2'] + row['DH3']) / 3
    else:
        score = (row['DH1'] + row['DH2'] + row['DH3']) / 3

    return 1 if score >= 5 else 0

df['KQXT'] = df.apply(kqxt, axis=1)
```

```python
df.to_csv("processed_dulieuxettuyendaihoc.csv", index=False)
print("Đã lưu file thành công!")
```
